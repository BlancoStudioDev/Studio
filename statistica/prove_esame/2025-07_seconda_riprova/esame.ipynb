### Esercizio 1
Considerate la funzione definita da
$$
f(x) = (m-x) \dfrac{a}{m} \, I_{[0,m]}(x),
$$
dove $a,m \in \mathbb R^+$.

1. Ricavate il valore di $a$, espresso in funzione di $m$, che rende $f$ una densità di probabilità.


    D'ora in avanti, fisseremo $\alpha$ al valore determinato al punto precedente.


Per ricavare $\alpha$ dobbiamo andare a dichiarare $f(x)$ come una funzione di densità di probabilità, quindi banalmente andandone
a fare l'integrale.

$$
f(x) = (m-x) \dfrac{a}{m} \, I_{[0,m]}(x) 
$$
$$
F(x) = \int(m-x) \dfrac{a}{m} = \int_0^m m*\frac{a}{m} - x*\frac{a}{m} = \int_0^m m*\frac{a}{m} - \int_0^m x*\frac{a}{m} =
$$
$$
= [a*x]_0^m - [\frac{x^2}{2} * \frac{a}{m}]_0^m = a*m - \frac{a*m^2}{2m} = \frac{a*m}{2}
$$

Ora noi sappiamo che se questa cosa la poniamo uguale a 1 allora otteniamo il valore di $\alpha$ espresso in funzione di m:

$$
\frac{a*m}{2} = 1
$$

quindi a:

$$
a = \frac{2}{m}
$$
$$
f(x) = \frac{2}{m} - \frac{2x}{m^2}
$$

2. Sia $X$ una variabile aleatoria la cui distribuzione è descritta dalla densità $f$. Ricavate il valore atteso di $X$, esprimendolo in funzione di $m$.

Per determinare il valore atteso di questa variabile aleatoria $X$ possiamo andare a scrivere che:

$$
E[X] = \int_0^m xf(x) = \frac{a}{m}\int_0^m x(m-x) = \frac{a}{m}\int_0^m m*x-x^2 = \frac{a}{m}[m*\frac{x^2}{2} - \frac{x^3}{3}]_0^m = 
$$
$$
= \frac{m^3}{2} - \frac{m^3}{3} = \frac{3*m^3 - 2*m^3}{6} = \frac{m^3}{6} * \frac{a}{m} = \frac{m^3}{6}*\frac{2}{m^2} = \frac{m}{3}
$$

3. Ricavate la varianza e la deviazione standard di $X$, esprimendole in funzione di $m$.

Per ricavare la varianza, si applica lo stesso metodo utilizzato precedentemente, solamente andando a moltiplicare per $x^2$ e non per $x$
e basta:

$$
E[X^2] = \int_0^m x^2f(x) = \frac{a}{m}\int_0^m x^2(m-x) = \frac{a}{m}\int_0^m x^2 * m - x^3 = \frac{a}{m} * [\frac{x^3}{3}*m - \frac{x^4}{4}]_0^m =
$$
$$
= \frac{m^4}{3} - \frac{m^4}{4} = \frac{4m^4 - 3m^4}{12} = \frac{m^4}{12} * \frac{a}{m} = \frac{m^4}{12} * \frac{2}{m^2} = \frac{m^2}{6}
$$
$$
Var(X) = E[X^2] - (E[X])^2 = \frac{m^2}{6} - (\frac{m}{3})^2 = \frac{m^2}{6} - \frac{m^2}{9} = \frac{m^2}{18}
$$

Per calcolare la deviazione standard andiamo a fare la radice quadrata della varianza:

$$
\sigma_X = \sqrt{\frac{m^2}{18}} = \frac{m}{3 * \sqrt{2}}
$$

4. Ricavate la forma analitica della funzione di ripartizione di $X$, che indicheremo con $F$, esprimendola in funzione di $m$.

Per ricavare la forma analitica della funzione di ripartizione andiamo ad integrare utilizzando però come variabile di integrazione t:

$$
F(T) = \int_0^x f(t)dt = \frac{2}{m^2}\int_0^x m - t dt = \frac{2}{m^2} * [mt - \frac{t^2}{2}]_0^x = \frac{2}{m^2} * [m * x - \frac{x^2}{2}] = 
$$
$$
= \frac{2x}{m} - \frac{x^2}{m^2}
$$

6. Sia $Y$ una variabile aleatoria distribuita secondo il modello esponenziale. Ricavate il valore del suo parametro $\lambda$, esprimendolo in funzione di $m$ e in modo che risulti $\mathbb E[X] = \mathbb E[Y]$.


Il valore atteso di una variabile aleatoria esponenziale è descritto come:

$$
E[Y] = \frac{1}{\lambda}
$$

quindi possiamo dire che, siccome: $E[Y] = E[X]$, $E[X] = \frac{1}{\lambda} = \frac{m}{3}$, \lamba sarà quindi uguale a:

$$
\frac{1}{\lambda} = \frac{m}{3}
$$
$$
\lambda = \frac{3}{m}
$$

7. Calcolate la varianza di $Y$, esprimendola in funzione di $m$. Calcolate poi il rapporto $\text{Var}(Y)/\text{Var}(X)$, commentando il risultato ottenuto.


La varianza in un modello esponenziale è definita come: $\frac{1}{\lambda^2}$, quindi basta sostituire: $\lambda = \frac{3}{m}$ nel calcolo:

$$
Var(Y) = \frac{1}{\lambda^2} = \frac{1}{\frac{9}{m^2}} = \frac{m^2}{9}
$$

Il rapporto tra le varianze è abbastanza semplice da calcolare:

$$
\frac{\frac{m^2}{9}}{\frac{m^2}{18}} = \frac{18m^2}{9m^2} = 2
$$

Dato che il rapporto viene 2 questo vorrà dire che i dati sono distribuiti con due volte la varianza, quindi la dispersione nello spazio
rispetto a X
